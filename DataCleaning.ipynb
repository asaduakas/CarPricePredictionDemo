```python
import os, pandas as pd
from pathlib import Path
```

# The DVM-CAR Dataset
The data used in this project consists of 2 parts:
### Part I: Image Data 
- 1,451,784 images from 899 UK market car models. 
https://deepvisualmarketing.github.io/
- 300x300 resolution, with background removed.
- Here, a 730MB subset with quality checked front-view images only was used, instead of entire 13.6GB.
### Part II: Tabular Data
- More specifically, the Ad and Image tables were joined.
- The Ad table contains advertisement info and the Image table contains the corresponding image from the advertisement

```python
TABLE_DIR = r"~/car-price-demo/tables"
IMG_ROOT = r"~/car-price-demo/images/confirmed_fronts"
PROJECT_ROOT = r"~/car-price-demo"

ad_table = pd.read_csv(os.path.join(TABLE_DIR, "Ad_table.csv"))
img_table = pd.read_csv(os.path.join(TABLE_DIR, "Image_table.csv"))

expandedImgRoot = os.path.expanduser(os.path.expandvars(IMG_ROOT))
path_to_img = Path(expandedImgRoot).resolve(strict=False)

expandedProjectRoot = os.path.expanduser(os.path.expandvars(PROJECT_ROOT))
path_to_project = Path(expandedProjectRoot).resolve(strict=False)
```

```python
tar_ad_ids = set(ad_table['Adv_ID'])
ad_table_w_img = ad_table.copy()

id_to_file = {}

for root, dirs, files in os.walk(path_to_img, topdown=True, followlinks=True):
    for file in files:
        adv_id = "$$".join(file.split("$$")[4:6])
        full_path = os.path.join(root, file)
        rel_path = os.path.relpath(full_path, start=path_to_project)
        id_to_file[adv_id] = rel_path

ad_table_w_img["FilePath"] = ad_table_w_img["Adv_ID"].map(id_to_file)

ad_table_w_img_clean = ad_table_w_img.dropna(subset=["FilePath"])
ad_table_w_img_clean.reset_index(drop=True, inplace=True)

ad_table_w_img_clean.to_csv("Ad_table_with_files.csv", index=False)

print("Cleaned table saved! Number of rows:", len(ad_table_w_img_clean))
```
